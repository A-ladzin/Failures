```python
from textgrid import TextGrid
from pydub import AudioSegment

# Load audio and TextGrid
audio = AudioSegment.from_wav("RuDevices/0/0/02d1e95f-e160-4fc3-be5c-b5ea5136267c.wav")
tg = TextGrid.fromFile("aligned0/02d1e95f-e160-4fc3-be5c-b5ea5136267c.TextGrid")

# Choose the tier name (e.g., "words", "phones")
tier = tg.getFirst("words")  # Change if needed

# Export each interval
for i, interval in enumerate(tier):
    if interval.mark.strip():  # Skip silences
        start_ms = int(interval.minTime * 1000)
        end_ms = int(interval.maxTime * 1000)
        segment = audio[start_ms:end_ms]
        label = interval.mark.strip().replace(" ", "_")
        segment.export(f"segment_{i}_{label}.wav", format="wav")
```
